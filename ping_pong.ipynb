# Ping Pong Örnek Kodu

Basit, konsol tabanlı bir ping pong (masa tenisi skor simülasyonu) örneği.

```python
import random

def ping_pong(win_score=5):
    player_score = 0
    computer_score = 0

    print("Ping Pong basladi! Ilk 5 sayiyi alan kazanir.\n")

    while player_score < win_score and computer_score < win_score:
        point = random.choice(["player", "computer"])
        if point == "player":
            player_score += 1
            print(f"Ping! Sen sayi aldin -> Sen: {player_score} | Bilgisayar: {computer_score}")
        else:
            computer_score += 1
            print(f"Pong! Bilgisayar sayi aldi -> Sen: {player_score} | Bilgisayar: {computer_score}")

    print("\nOyun bitti!")
    if player_score > computer_score:
        print("Kazanan: Sen")
    else:
        print("Kazanan: Bilgisayar")

ping_pong()
```
